```python
import math

# Fungsi yang dihampiri
def f(x):
    return math.sqrt(x)

# Menghitung koefisien Taylor
def koefisien(a, k):
    hasil = math.sqrt(a)

    for i in range(1, k + 1):
        hasil *= (0.5 - (i - 1)) / (i * a)

    return hasil

# Titik pusat dan titik hampiran
a = 25
x = float(input("Masukkan nilai x: "))
h = x - a

# Nilai sebenarnya
asli = f(x)
P = 0

print("\n=== POLINOMIAL TAYLOR ===")
print(f"Nilai sebenarnya = {asli:.10f}")

# Menghitung P0 sampai P4
for k in range(5):
    c = koefisien(a, k)
    P += c * h**k
    galat = abs(asli - P)

    print(f"\nP{k} = {P:.10f}")
    print(f"Koefisien = {c:.10f}")
    print(f"Galat absolut = {galat:.10e}")

# Mencari orde minimum
P = 0
print("\n=== ORDE MINIMUM ===")

for k in range(91):
    c = koefisien(a, k)
    P += c * h**k
    galat = abs(asli - P)

    if galat < 1e-6:
        print(f"Orde minimum = {k}")
        print(f"Galat = {galat:.10e}")
        break
else:
    print("Orde minimum tidak ditemukan sampai 90.")
```
